#### Why "seed"?

- It's a metaphor from gardening: plant the same seed, get the same plant
- they use a pseudorandom number generator (PRNG) — an algorithm that produces a long sequence of numbers that look random (well-distributed, unpredictable-looking) but are actually generated by a deterministic formula.

```python
random.seed(49)
print(random.random())  
print(random.random())  

random.seed(49)  # reset to the same seed
print(random.random())  

```

#### Random.gauss
Returns a random float from a Gaussian (normal) distribution, given a mean (mu) and standard deviation (sigma).

```python
#random.gauss(mu ,sigma)
random.gauss(0,1)   ## takes mu and sigma to return the value
```

```python
# Generate several values clustered around a mean of 0
samples = [random.gauss(0, 1) for _ in range(5)]
print(samples)
```

#### random.getstate()

- Captures the internal state of the random number generator at that moment, so you can save it and return to it later with random.setstate()

```python
# random.getstate()
```

```python
random.seed(11)  ##sets a starting point
state = random.getstate()  # save current state

print(random.random())  # some value, e.g. 0.5714025946899135
print(random.random())  # a different value

random.setstate(state)   # restore to the saved point
print(random.random())   # 0.5714025946899135 (repeats!)
```

sys.setrecursionlimit(n)

Changes the maximum depth of recursive function calls Python allows (default is usually 1000). Useful for deep recursion, but risky — too high can crash your program (stack overflow).

```python
sys.setrecursionlimit(3000)

def recurse(n):
    if n == 0:
        return 0
    return recurse(n - 1)

print(recurse(2900))  # works now; would've failed with default limit
```

```python
sys.getrecursionlimit()     #prints the recursion limit
```

sys.exc_info()

Returns information about the exception currently being handled — a tuple of (type, value, traceback). Used inside except blocks for detailed error inspection.

```python
try:
    1 / 0
except ZeroDivisionError:
    exc_type, exc_value, exc_tb = sys.exc_info()
    print("Type:", exc_type)     # <class 'ZeroDivisionError'>
    print("Value:", exc_value)   # division by 
```
