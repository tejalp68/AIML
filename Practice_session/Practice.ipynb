# Q1 — Even or Odd Counter
- 20 min · loops, conditionals, comprehensions
- You're given a hardcoded list of numbers:
- numbers = [12, 7, 9, 4, 22, 15, 8, 3, 30, 11]
- Write a script that:
- Loops through the list and prints whether each number is even or odd
- Counts how many are even and how many are odd, using simple counters
- Uses a comprehension to build a new list containing only the even numbers
- Prints the final counts and the even-numbers list

```python
numbers = [12, 7, 9, 4, 22, 15, 8, 3, 30, 11]

e_count = 0
o_count = 0

for i in numbers:
    if (i % 2 == 0):
        print(i,"Even Number")
        e_count += 1
    else:
        print(i ,"Odd Number")
        o_count += 1


print(f"Even number count is : {e_count}")
print(f"Odd number count is : {o_count}")


        
even =[n for n in numbers if n % 2 ==0]
print("Only even numbers :" ,even)
```

```python
def calculate_average(numbers: list) -> float :

    """
    Calculate the average (of a list of numbers.

    Args:
        numbers (list): A list of numeric values 

    Returns:
        float: The average of the numbers in the list
    """

    avg = sum(n for n in numbers) / len(numbers)
    return avg

numbers:list =[1,2,5,7]
print(calculate_average(numbers))



# driver code
list1 =[10,20,30]
list2=[]
def test_case(list):
    try :
        avg = calculate_average(list)
        print(avg)
    except ZeroDivisionError:
        print(f"Couldn't calculate average of {list} — because list is empty!")
test_case(list1)
test_case(list2)
```

# Q4 — Student Class
- 30 min · classes, __init__, type hints, one dunder method
- Design a Student class:
- __ init__(self, name: str, marks: list) — both type-hinted instance attributes
- A method average(self) -> float that returns the average of the student's marks
- __ str__ — returns a readable one-line summary, e.g. "Sakshi: average 82.5"
- Then create a hardcoded list of 3 Student objects, print each one (this should use your __str__ automatically), and print the name of
whichever student has the highest average

```python
class Student :
    def __init__(self , name: str, marks: list):
        self.name = name
        self.marks = marks
    def average(self) -> float:
        avg = sum(self. marks) /len(self.marks)
        return avg
    def __str__(self) -> str:
        return f"{self.name}: average {self.average()}"    #calling average function here




students = [
    Student("Sakshi", [90, 85, 75, 80]),
    Student("Rahul", [60, 70, 65, 55]),
    Student("Meera", [95, 92, 98, 91])
]


for student in students:
    print(student)
    
topper = students[0]
for student in students:
    if student.average() > topper.average():
        topper = student

print(f"\nTopper: {topper.name} with average {topper.average():.1f}")
    
```

# Q5 — Import and Use a Module
- 20 min · import mechanics, module functions
- Using only what's been shown in class (the random, math, and sys modules — pick any one or combine them):
- Write a short script (10-15 lines) that imports at least one module and uses at least two different functions from it
- Example ideas: a simple dice-roll-until-you-get-a-6 counter (random), a script that computes the area and circumference of a circle given a
 radius (math), or a script that prints the Python version and exits cleanly with a message (sys)
- Add a short comment above each function call explaining what it does


```python
import math

radius = int(input("Enter a number"))

# math.pi gives an accurate value of π for our calculations
area = math.pi * radius ** 2

# math.pi is reused here to compute the circle's circumference
circumference = 2 * math.pi * radius

print(f"Radius: {radius}")
print(f"Area: {area:.2f}")
print(f"Circumference: {circumference:.2f}")

# math.sqrt() returns the square root — used here to find the
# "equivalent side" of a square with the same area as our circle
equ_side = math.sqrt(area)
print(f"A square with the same area would have sides of length: {equ_side:.2f}")
```
