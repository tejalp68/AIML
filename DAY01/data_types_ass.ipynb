### Dictionary 

```python
# Create a dictionary
student = {
    "name": "Alice",
    "age": 21,
    "course": "Computer Science",
    "grade": "A"
}
```

```python
print("Name:", student["name"])
print("Age (using get):", student.get("age"))
```

```python
#Add a new key-value pair
student["email"] = "alice@example.com"
print(student)
```

```python
# Update an existing value
student["grade"] = "A+"
print(student)
```

```python
# 4. Remove a key-value pair
student.pop("course")
```

```python
#Check if a key exists
print("Has 'name'?", "name" in student)
```

```python
#Loop through keys and values
for key, value in student.items():
    print(f"{key}: {value}")
```

```python
# Get all keys and values
print("Keys:", list(student.keys()))
print("Values:", list(student.values()))
```

### Set

```python
fruits = {"apple", "banana", "cherry", "mango"}

fruits.add("orange")
#Add multiple elements
fruits.update(["grape", "kiwi"])
print(fruits)
```

```python
# Remove an element (raises error if not found)
fruits.remove("banana")
print(fruits)
```

```python
#Discard an element (no error if not found)
fruits.discard("pineapple")
print(fruits)
```

```python
print("Is 'apple' in fruits?", "apple" in fruits)
```

```python
print("Number of items:", len(fruits))
```

```python
# Loop through set
for fruit in fruits:
    print(fruit)
```
