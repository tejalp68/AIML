### **__init__()** method

 - obj creation process is repeated then this method is used



```python
class Student :
    def __init__(self,name,age):
        self.name = name
        self.age = age
    def display_value(self):
        print(self.name)
        print(self.age)
stud1 = Student('Tejal',21)
stud1.display_value()
```

#### Methods : a function that lives in a class 

```python
class Student:
    def __init__(self,name,age):
        self.name = name
        self.age = age
    def introduce(self):
        print(f'My name is {self.name}')
        print(f'My age is {self.age}')
stud1 = Student('Tejal','21')
stud1.introduce()   # method
stud2 = Student('Rishi' ,'16')
stud2.introduce()   # python automatically passes student in as a self  
```

### instance attribute vs class attribute
- instance attribute  - belongs to one object 
- class attribute - shared by every object .set directly inside class

```python
class BankAccount:
    def __init__(self,owner,balance):
        self.owner =owner
        self.balance =balance
    def deposit(self,amount):
        self.balance += amount
    def withdraw(self,amount):
        if amount <= self.balance :
            self.balance -= amount
        else:
            print("Insufficient balance")
    def show_balance(self):
        print(f"Balance is :{self.balance}" )
account =BankAccount('Tejal',5000)
account.deposit(1500)
account.withdraw(1000)
account.show_balance()
```

#### Dunder methods 
- __ init __()
- __ str __()
- __ len __()
- __ eq __()

```python
class Student:
    def __init__(self,name,age):
        self.name = name
        self.age = age
    def __str__(self):
        return f'{self.name}'
    def __repr__(self):    #used for debugging and tell what data type is given
        return f'Student({self.age!r})'

stud1 = Student('Tejal',20)
print(stud1)
print(repr(stud1))
```
