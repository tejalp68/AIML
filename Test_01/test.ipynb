### Q2 (Documentation + Execute)
- Docs: https://pytorch.org/tutorials/beginner/basics/buildmodel_tutorial.html
- Create a small network with two nn.Linear layers: nn.Linear(3, 5) followed by nn.Linear(5, 1) . Pass a random tensor
of shape (1, 3), using torch.rand , through it and print the final output shape and value

```python
import torch
import torch.nn as nn

model = nn.Sequential(
    nn.Linear(3, 5),
    nn.Linear(5, 1)
)

x = torch.rand(1, 3)
output = model(x)

print("Output shape:", output.shape)
print("Output value:", output)
```
