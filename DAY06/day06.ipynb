## Iterators

- An iterator is any object that follows the iterator protocol — it knows how to give you one value at a time, and knows when it's done.

- Two methods make something an iterator:

  - __ iter__(self) → returns the iterator object itself
  - __ next__(self) → returns the next value, or raises StopIteration when exhausted
    

- two roles:
 - iterable = has __ iter__ , can produce an iterator when asked
 - iterator = has __ next__ , actually produces values one at a time and remembers position

````python
nums = [10, 20, 30]

print(type(nums))                  # <class 'list'>
print(hasattr(nums, '__iter__'))   # True -> list has __iter__ defined
print(hasattr(nums, '__next__'))   # False -> list itself is NOT an iterator
````

````python
s = "hello"
d = {"a": 1}
t = (1, 2, 3)

print(hasattr(s, '__iter__'))   # True # String
print(hasattr(d, '__iter__'))   # True # Dict
print(hasattr(t, '__iter__'))   # True # Tuple
````

````python
nums = [10, 20, 30]

it_a = iter(nums)              # using the built-in function
it_b = nums.__iter__()         # calling the dunder method directly

print(type(it_a))              # <class 'list_iterator'>
print(type(it_b))              # <class 'list_iterator'>
print(next(it_a))              # 10
print(next(it_b))              # 10  <- both behave identically, proves iter() == calling __iter__()
````

````python
nums = [10, 20, 30]
it = iter(nums)

print(id(nums))      # some address, e.g. 140234529341632
print(id(it))        # a totally different address, e.g. 140234529298112
print(nums is it)    # False -> definitely two different objects
print(type(nums))    # <class 'list'>
print(type(it))      # <class 'list_iterator'>
````

````python
it1 = iter(nums)
it2 = iter(nums)
print(it1 is it2)    # False -> fresh object each call
````

````python
nums = [10, 20, 30]
it = iter(nums)

print(next(it))     # 10
nums.append(40)     # mutate the ORIGINAL list
print(next(it))     # 20
print(next(it))     # 30
print(next(it))     # 40  <- iterator sees the change, proves it holds a reference, not a copy
````

````python
nums = [1, 2, 3]
print(nums is iter(nums))   # False -> list is NOT its own iterator

c = Countdown(3)
print(c is iter(c))          # True -> Countdown IS its own iterator
````

````python
class Countdown:
    def __init__(self, start):
        self.current = start

    def __iter__(self):
        return self  # I am my own iterator

    def __next__(self):
        if self.current <= 0:
            raise StopIteration
        val = self.current
        self.current -= 1
        return val

c = Countdown(3)
it = iter(c)      # calls c.__iter__() -> returns c itself
print(next(it))    # calls c.__next__() -> 3
print(next(it))    # 2
print(next(it))    # 1
#print(next(it))   # raises StopIteration
````

````python
class CountUpto:
    def __init__(self,limit):
        self.limit =limit
        self.n =0
    def __iter__(self):
        return self       # i am my own iterator
        
    def __next__(self):
        if self.n >= self.limit :
            raise StopIteration
        self.n += 1
        return self.n
        
for x in CountUpto(10):
    print(x)
    
````

## Generator

- Generators are a shortcut for writing iterators using functions instead of classes. Any function with yield in it becomes a generator function — calling it doesn't run the code, it returns a generator object (which automatically implements __ iter__ and __ next__ for you).

````python
def count_up_to(limit):
    n = 0
    while n < limit:
        n += 1
        yield n

for x in count_up_to(3):
    print(x)   # 1 2 3
````

- yield pauses the function, remembers everything (local variables, instruction pointer), and resumes exactly there on the next next() call

## Decorators

- A decorator is a function that wraps another function (or class) to add behavior without modifying its source code. This relies on the fact that in Python, functions are first-class objects — they can be passed around, returned, and defined inside other functions (closures).

````python
def logger(func):
    def wrapper(*args, **kwargs):
        print(f"Calling {func.__name__}")
        result = func(*args, **kwargs)
        print(f"Done with {func.__name__}")
        return result
    return wrapper

                        # also can be written as @logger
def greet(name):
    return f"Hello, {name}"
    
greet = logger(greet) 
greet("Sam")
````

- Iterator protocol (manual, verbose)
-    → Generators (yield keyword auto-implements the protocol, adds laziness)
-         → Context manager protocol (manual, verbose: __enter__/__exit__)
-             → @contextmanager (decorator + generator = auto-implements the protocol)
- Decorators (closures) are the connective tissue enabling all these transformations
