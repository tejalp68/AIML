```python
# Global scope variable
message = "Hello from global scope"

def demo_scope():
    # Local scope variable
    local_var = "Hello from local scope"

    # Reading a global variable
    print(message)
    print(local_var)

    # This variable only exists inside the function
    local_only = "I disappear after the function ends"
    print(local_only)


# Calling the function
demo_scope()

print(message)   # Works — message is global
```
